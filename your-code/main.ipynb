```python
# Invert values
my_list = [1, 2, 3, 4, 5]
my_list1 = [1, -2, 3, -4, 5]
my_list2 = []

# solution
def inverter(list_):

  inverted_list = [ -inv_val for inv_val in list_ ]

  return inverted_list

inverter(my_list)
inverter(my_list1)
inverter(my_list2)
```

```python
# Sum square n 
my_list = [1, 2, 3]
my_list1 = [-4, 5]

# solution
def sum_sqr(list_):

  sqr_list = [ val ** 2  for val in list_ ]

  return sum(sqr_list)

sum_sqr(my_list)
sum_sqr(my_list1)
```

```python
#Digitize
int_1 = 1234
int_12 = 567

def digitizer(my_int):

  dig_ls = [ int(digit) for digit in str(my_int) ]

  return dig_ls

digitizer(int_1)
```

```python
# list filtering

list_1 = [1,2,5, 'asd', 7, 'ff']

def no_stringer(my_list):

  no_str_ls = [ only_int for only_int in my_list if type(only_int) == int ]

  return no_str_ls

no_stringer(list_1)
```

```python
# Multiples of 3 or 5

def mult_sum(num):

  list_mults = [ i if i >= 0 and (i % 3 == 0 or i % 5 == 0) else 0 for i in range(num) ]

  return sum(list_mults)

mult_sum(16)
```

```python
# Count of positive / sum of negatives

def pos_neg(my_array):

  positives = []

  positives.extend([i for i in my_array if i > 0])
  countpos = len(positives)

  sumneg = sum([i for i in my_array if i < 0])

  return [countpos,sumneg]

list_1 = [1,2,5, 7, 0, -2, -20]
pos_neg(list_1)
```

```python
# Categorize new member

input_list = [[18, 20], [45, 2], [61, 12], [37, 6], [21, 21], [78, 9]]


output = ["Open", "Open", "Senior", "Open", "Open", "Senior"]

def seniority(input_):

  output_ = [ "Senior" if input_[i][0] >= 55 and input_[i][1] > 7 else "Open" for i in range(len(input_))]

  return output_


seniority(input_list)

```
